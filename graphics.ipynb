```python
import matplotlib.pyplot as plt
import numpy as np

# Названия моделей
models = [
    "Линейная регрессия",
    "Random Forest",
    "Gradient Boosting"
]

# Значения метрик
mae = [6.2, 4.9, 3.8, 3.5]
rmse = [8.7, 6.4, 5.2, 4.9]
r2 = [0.71, 0.83, 0.89, 0.91]

# Позиции на оси X
x = np.arange(len(models))
width = 0.25  # ширина столбиков

# Создаём фигуру
plt.figure(figsize=(10, 6))
plt.bar(x - width, mae, width, label='MAE ↓')
plt.bar(x, rmse, width, label='RMSE ↓')
plt.bar(x + width, r2, width, label='R² ↑')

# Настройки осей
plt.xticks(x, models, rotation=15, ha='right')
plt.title("Сравнение производительности моделей", fontsize=14, fontweight='bold')
plt.ylabel("Значение метрики")
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Добавим подписи над столбиками
def add_labels(values, xpos):
    for i, v in enumerate(values):
        plt.text(i + xpos, v + 0.05, f"{v:.2f}", ha='center', fontsize=9)

add_labels(mae, -width)
add_labels(rmse, 0)
add_labels(r2, width)

plt.tight_layout()
plt.show()
```

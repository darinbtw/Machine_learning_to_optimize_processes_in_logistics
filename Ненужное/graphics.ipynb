```python
import matplotlib.pyplot as plt
import numpy as np

# Названия моделей
models = [
    "Линейная регрессия (Baseline)",
    "Random Forest Regressor",
    "XGBoost Regressor"
]

# Значения метрик для каждой модели
# ВАЖНО: Подставлены ВАШИ фактические результаты
# MAE (минуты)
mae_values = [20.04, 17.97, 17.81]
# RMSE (минуты)
rmse_values = [31.60, 23.56, 23.00]
# R² (доля, не проценты)
r2_values = [0.4374, 0.7889, 0.7988] 

x = np.arange(len(models)) # Позиции для моделей
width = 0.25 # Ширина каждого столбика

plt.figure(figsize=(14, 7))

# Функция для построения столбцов с подписями
def plot_bars(data, offset, label, color, format_spec):
    bars = plt.bar(x + offset, data, width, label=label, color=color)
    for i, bar in enumerate(bars):
        yval = bar.get_height()
        # Специальное форматирование для R²
        if label.startswith('R²'):
            # R^2 отображаем в виде процентов с одним знаком после запятой
            text_label = f"{yval * 100:.1f}%"
        else:
            text_label = f"{yval:{format_spec}}"
            
        # Установка позиции подписи выше столбика
        plt.text(bar.get_x() + bar.get_width()/2, yval + (max(rmse_values) * 0.02), text_label, 
                 ha='center', va='bottom', fontsize=9, fontweight='bold')
    return bars

# Построение графиков
plot_bars(mae_values, -width, 'MAE (мин)', '#007ACC', '.2f')
plot_bars(rmse_values, 0, 'RMSE (мин)', "#BD07FF", '.2f')
plot_bars(r2_values, width, 'R² (точность)', '#4CAF50', '.4f') # R² отображается в диапазоне [0, 1]

# Настройки
plt.xticks(x, models, rotation=0, ha='center', fontsize=11)
plt.title("Сравнительный анализ производительности регрессионных моделей", 
          fontsize=16, fontweight='bold', pad=20)
plt.ylabel("Значение метрики", fontsize=12)
plt.legend(loc='upper left', bbox_to_anchor=(0.01, 1), ncol=3)
plt.grid(axis='y', linestyle='--', alpha=0.6)

# Установка пределов оси Y для лучшей визуализации MAE и RMSE (R² будет внизу)
plt.ylim(0, max(rmse_values) * 1.1)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Метрики
mae_val = 17.97
rmse_val = 23.56
r2_val = 0.7889

# Переводим R² в проценты для лучшей визуализации и добавляем в список
# R^2 будет в процентах (0.7889 -> 78.89), но мы будем отображать его в диапазоне [0, 1] 
# или масштабируем MAE и RMSE, чтобы они были в одном диапазоне.

# Для наглядности, давайте представим только две метрики ошибки (MAE, RMSE) 
# и одну метрику точности (R²).
metric_values = [mae_val, rmse_val, r2_val]
metric_names = ['MAE (мин.)', 'RMSE (мин.)', 'R² (точность)']
colors = ['#4CAF50', '#FFC107', '#2196F3'] # Зеленый, желтый, синий

# Позиции на оси X
x = np.arange(len(metric_names))
width = 0.60 # Ширина столбиков

# Создаём фигуру
plt.figure(figsize=(8, 6))

# Построение гистограммы
bars = plt.bar(x, metric_values, width, color=colors)

# Настройки осей и заголовков
plt.xticks(x, metric_names, fontsize=12)
plt.title(
    "Оценка производительности модели Random Forest Regressor", 
    fontsize=14, 
    fontweight='bold', 
    pad=20
)
plt.ylabel("Значение метрики", fontsize=11)
plt.xlabel("Метрика", fontsize=11)
plt.grid(axis='y', linestyle='--', alpha=0.6)

# Добавим подписи над столбиками
def add_labels(bars):
    for bar in bars:
        # Получаем высоту (значение) столбика
        yval = bar.get_height()
        # Форматирование: если значение близко к 1 (R²), показываем 4 знака
        if yval < 1:
             label = f"{yval:.4f}"
        else:
             label = f"{yval:.2f}"
             
        plt.text(
            bar.get_x() + bar.get_width()/2, # Центр столбика по X
            yval + 1.5,                       # Над столбиком по Y
            label, 
            ha='center', 
            fontsize=10,
            fontweight='bold'
        )

add_labels(bars)

# Устанавливаем лимит Y, чтобы подписи не выходили за рамки
plt.ylim(0, max(metric_values) * 1.1 + 3)

plt.tight_layout()
plt.show()
```
